**Neural Networks** - The Advanced Learning Algorithms

**Biological Brain Model - Neurons and the brain**
- Neural Networks were first introduced to mimic human brain structure in software
- A neuron in the bassic unit of the brain and has dendrites(input wire), a soma(the central processing of the electrical input), and axon(output wire). Neurons communicate in a network and that is what leads to thought and intelligence
- Artificial Neural Networks work to mimic the basic structure of the neuron. They take inputs as numerical values, process it through some activation/accumulation function, and output to the next neurons in a network


**Neural Network Structure**
- A neural network is split into three components(Input Layer, Hidden Layer, and Output Layer)
- A layer is a group of neurons that connect to the previous layer and the next layer. Each neuron in the layer is connected to all the previous neurons in the previous layer and all the next neurons in the next layer
- The input layer takes a vector x representing the inputs from the training set and pass that into each of the neurons. The outputs of each neuron in this layer go to the hidden layer.
- There can be multiple hidden layers but they can be perceived as taking the inputs and creating their own features.
- The output layers takes inputs from the final hidden layer and output the prediction of the network

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import random as rand
```

```python
# Neuron Class Definition

class Neuron():
    def __init__(self, weights, bias, output):
        self.w = weights
        self.b = bias
        self.output = output
    
    def _sigmoid(self, z):
        return 1/(1 + np.exp(-z))
    
    def activation(self, input):
        z = np.dot(input, self.w) + self.b
        self.output = self._sigmoid(z)

# Test Neuron Model

n1 = Neuron(np.array([1, 2, 1]), 3, 0)

n1.activation(np.array([2, 3, 4]))

print(n1.output)

```

```python
# Layer of Neurons Defition which utilizes list of Neurons

class Layer():
    def __init__(self, input_a, list_neurons):
        self.input_a = input_a
        self.list_neurons = list_neurons
        self.a_vector = np.zeros(len(input_a))
    
    def compute_activation(self):
        for i in range(len(self.list_neurons)):
            self.list_neurons[i].activation(self.input_a)
            self.a_vector[i] = self.list_neurons[i].output

# Testing values
n1 = Neuron(np.array([0.5, 0.2, 0.1]), 0.3, 0)
n2 = Neuron(np.array([0.2, 0.1, 0.5]), 0.4, 0)
n3 = Neuron(np.array([0.1, 0.4, 0.8]), 0.5, 0)

new_layer = Layer(np.array([2, 3, 4]), [n1, n2, n3])

new_layer.compute_activation()

print(new_layer.a_vector)
```

**Complete Implementation of Basic Vectorized Neural Network with Feed Forward Algorithm**

```python
# Neuron Array representation: weight, bias
# The single neuron model does not scale well and we need to initialize objects for each neuron, making it an array representation is easier

class Layer_vector():
    def __init__(self, list_neurons):
        self.list_neurons = list_neurons
        self.a_vector = []
    
    def compute_activation(self, input_a):
        self.a_vector = np.zeros(len(self.list_neurons))
        for i in range(len(self.list_neurons)):
            self.a_vector[i] = 1/(1 + np.exp(-(np.dot(input_a, self.list_neurons[i][0]) + self.list_neurons[i][1])))
        return self.a_vector
# Feedforware Implementation

def sigmoid(z):
    return 1/(1+ np.exp(-z))

def feed_forward(input_a0, layer1, layer2, output_neuron_w, output_neuron_b):
    a_1 = layer1.compute_activation(input_a0)

    a_2 = layer2.compute_activation(a_1)

    output = sigmoid(np.dot(a_2, output_neuron_w) + output_neuron_b)
    return output


input_a0 = [1, 2, 4, 3]
layer1 = [
        [[0.5, 0.3, 0.1, 0.4], 0.3],
        [[0.4, 0.2, 0.6, 0.2], 0.1],
        [[0.6, 0.1, 0.8, 0.3], 0.2],
        [[0.1, 0.8, 0.8, 0.3], 1.8],
        [[0.9, 0.2, 0.3, 0.4], 1.5]
        ]

layer2 = [
        [[0.5, 0.1, 0.3, 0.1, 0.4], 0.2],
        [[0.4, 0.6, 0.2, 0.6, 0.2], 0.5],
        [[0.6, 0.4, 0.1, 0.8, 0.3], 0.3]
        ]

output_neuron_w = [0.4, 0.3, 0.6]
output_neuron_b = 1.8

Layer1 = Layer_vector(layer1)
Layer2 = Layer_vector(layer2)

prediction = feed_forward(input_a0, Layer1, Layer2, output_neuron_w, output_neuron_b)

print(prediction)
```

```python
# More Condensed Version of full pipeline training of neural network in tensorflow

layer1 = Dense(units = 3, activation='sigmoid')
layer2 = Dense(units = 1, activation='sigmoid')

model = Sequential([layer1, layer2])

model.compile(optimizer='adam', loss='binary_crossentropy')

x_train = np.array([[200, 17], [100,50], [50, 100], [300, 0]])
y_train = np.array([1,0,0,1])

model.fit(x_train, y_train)

x_new = np.array([30, 400])

model.predict(x_new)
```
