```python
# TensorFlow Implementation of Training

import tensorflow as tf
import numpy as np
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.losses import BinaryCrossentropy

model = Sequential([
    Dense(units=25, activation='sigmoid')
    Dense(units=15, activation='sigmoid')
    Dense(units=1, activation='sigmoid')
])

model.compile(loss=BinaryCrossentropy())

X = np.array([10, 20], [50, 20])
y = np.array([1], [0])

model.fit(X, y, epochs=100)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Three different activation functions: Sigmoid(logistic), ReLU, and linear(none)
# We choose the specific activation function based on the creteria of output. 
# If output is binary classification: Sigmoid for output layer
# If output is a value in [-inf, inf] then linear activation
# If output is a value in [0, inf] then ReLU is the best
# Usually the ReLU activation function is used for all hidden layers

def sigmoid(z):
    return 1/(1+ np.exp(-z))

def ReLU(z):
    z_copy = np.copy(z)
    z_copy[z_copy < 0] = 0
    return z_copy

def linear(z):
    return z

X_s = np.arange(-4,4,0.1)
y_sig = sigmoid(X_s)
y_ReLU = ReLU(X_s)
y_linear = linear(X_s)

plt.plot(X_s, y_sig, label='sigmoid')
plt.plot(X_s, y_ReLU, label='ReLU')
plt.plot(X_s, y_linear, label='linear')
plt.legend()
plt.show()

```
