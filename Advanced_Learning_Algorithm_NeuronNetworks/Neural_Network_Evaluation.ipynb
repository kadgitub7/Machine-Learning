**Methods to test Neural Networks**

We want to split the data into three different sets:
- Training set which is used to train the parameters of the model
- Validation set which is used to determine out of many differen trained models where you want to find which model type is best(which to pick based on lowest validation loss)
- Test set which is used to provide the user with a very good estimate on how well the model performs on data it has never seen. Usually conducted when we have already selected the optimal model

Bias is a measure of underfitting. High Bias is underfitting
Variance is a measure of overfitting. High Variance is overfitting

- You can identify High Bias if J_train is high and J_val is high
- You can identify High Variance if J_train is low, but J_val is high


Introducing a regulatization term can help to prevent overfitting. By introducing this term you add a threshold to the weights were you add the value of w to the loss so that it is generally directed to be lower.
- The lambda value needs to be tweaked so that it is not underfitting as well.
- If lambda is too large then the model only learns to keep weights low which might underfit
- If lambda is too small then the model has no regularization term and can overfit
- Finding the lambda where J_cv is minimized is the ideal regularization term

Utilizing a baseline level of performance is very important as that is what you can compare your results to.

If the problem is something where 0 % error is possible than that is your baseline else you can use Human Level performance, Other competing algorithm performance, or your experience to determine if the algorithm is doing well.

Understanding where your algorithm is getting wrong answers is also important. You can understand where errors are most commonly coming from and try to imporve the model architecture for instance to limit those errors further.

Data can often be a limiting factor. If you error analysis shows that the majority of errors originate from one subsection, then increasing the training examples in that set can help the algorithm learn more efficeiently.

You can do this trhough two main ways:
- Data augmentation takes existing data and alterns it(adds noise or distortions) to create new training examples which help the algorithm learn easily
- Data synthesis is where you originally come up with completely new training data not using past training data as a reference. This is a lot of work.

Transfer learning(pre-training + fine tuning) is a method which involves using the already trained weights of a large neural network for a similar input but different output task. For example if we want to recognize handwritten digits but there is a large NN for image classification, we can take the layers except for the last one and place our own output layer to take from that one. This way the general characteristics extracted from the previous algorithm can be useful for our version.

Precision = True positive / (True positive + False positive)
Recall    = True Positive / (True positive + False Negative)
