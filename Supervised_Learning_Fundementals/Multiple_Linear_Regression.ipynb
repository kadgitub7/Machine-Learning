Multiple Linear Regression Full Implementation with vectorization

```python
import random as rand
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Dataset initialization and printing
X_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])

df = pd.DataFrame(X_train, columns=['Size', '# of Bedrooms', '# of Floors', 'Age'])
df2 = pd.DataFrame(y_train, columns=['Price'])

print(df)
print(df2)
```

```python
# Initialize Weights and Bias values

w = np.zeros(4)
b = rand.randint(0,50)

print("Initialized values for Weights Vector(w) and Bias(b)")
print(f"w = {w}")
print(f"b = {b}")
```

```python
# Define Model Function

def f_wb(x, w, b):
    return np.dot(x,w) + b

print(f_wb(w, X_train[0], b))
```

```python
# Define a Cost Function

def J(x, y, w, b):
    error = f_wb(x, w, b) - y
    return (1/(2*len(x)))*sum(error**2)

print("Initial Cost From Weights and Bias")
print(X_train, y_train, w, b)
print(J(X_train, y_train, w, b))
```

```python
# Define Derivative and Gradient Descent Algorithm

def der_J_w(x, y, w, b):
    delta = 0.0001
    grads = np.zeros(len(w))
    for i in range(len(w)):
        w_plus = w.copy()
        w_plus[i] += delta
        grads[i] = (J(x,y, w_plus,b) - J(x, y, w, b))/delta
    print(grads)
    return grads

def der_J_b(x, y, w, b):
    delta = 0.0001
    der = (J(x,y, w,b+delta) - J(x,y, w,b))/delta
    print(der)
    return der

def grad_desc(x, y, w, b):
    alpha = 1e-7
    tmp_w = w - alpha*der_J_w(x, y, w, b)
    tmp_b = b - alpha*der_J_b(x, y, w, b)
    w = tmp_w
    b = tmp_b
    return w, b

w, b = grad_desc(X_train, y_train, w, b)
print(w, b)
```

```python
# Training Stage

epoch = 25
w = np.zeros(4)
b = rand.randint(0,50)

for i in range(len(w)):
    w[i] = rand.randint(0,10)

for i in range(epoch):
    w, b = grad_desc(X_train, y_train, w, b)
    print(f"Epoch {i} | Loss = {J(X_train, y_train, w, b)} | w = {w} | b = {b}")
```

**Optimization Techniques: Feature Scaling, Learning Rate, Feature Engineering**

**Feature Scaling Methods**
- Max Feature Scaling: Divides all values in a feature by the max value, restricts to [0, 1]
- Mean Feature Scaling: Computes mean and performs: (x - mu)/(max-min) to get values centered around [-1,1]
- Z scale normalization: calculated standard deviation and mean and normalizes by: x = (x - mu)/sigma

```python
# Max Feature Scaling

df = pd.DataFrame(X_train, columns=['Size', '# of Bedrooms', '# of Floors', 'Age'])
df2 = pd.DataFrame(y_train, columns=['Price'])

print("Unscaled values")
print(df)
print(df2)

print("Max Feature Scaling")
features_set = []
for i in range(len(X_train[0])):
    tmp = []
    for j in range(len(X_train)):
        tmp.append(X_train[j][i])
    features_set.append(tmp)

features_set = np.array(features_set)

max = np.zeros(len(features_set))
for i in range(len(features_set)):
    max[i] = features_set[i].max()
X_train_max = X_train.copy()
X_train_max = np.array(X_train_max, dtype='float')
for i in range(len(X_train_max[0])):
    for j in range(len(X_train_max)):
        X_train_max[j][i] = X_train_max[j][i] / max[i]

df = pd.DataFrame(X_train_max, columns=['Size', '# of Bedrooms', '# of Floors', 'Age'])
print(df)

# Mean Normalize
print("Mean Normalize")
mean = np.zeros(len(features_set))
for i in range(len(features_set)):
    mean[i] = features_set[i].mean()

min = np.zeros(len(features_set))
for i in range(len(features_set)):
    min[i] = features_set[i].min()

X_train_mean = X_train.copy()
X_train_mean = np.array(X_train_mean, dtype='float')
for i in range(len(X_train_mean[0])):
    for j in range(len(X_train_max)):
        X_train_mean[j][i] = (X_train_mean[j][i] - mean[i])/(max[i] - min[i])

df = pd.DataFrame(X_train_mean, columns=['Size', '# of Bedrooms', '# of Floors', 'Age'])
print(df)
```

Different Learning rates can lead to different convergences. You don't want it too small since that will take too long, but too big and it will overshoot.
